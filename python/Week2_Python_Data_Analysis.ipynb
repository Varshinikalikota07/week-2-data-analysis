```python

import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Anu", "Kiran", "Rahul"],
    "Age": [21, None, 23, 22, None],
    "Salary": [25000, 30000, None, 28000, 35000]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("\nData After Filling Missing Values:")
print(df)
```

```python

import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Ravi", "Anu", "Priya"],
    "Product": ["Mouse", "Keyboard", "Mouse", "Laptop", "Keyboard"],
    "Amount": [500, 1200, 500, 45000, 1200]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nData After Removing Duplicates:")
print(df)

print("\nNumber of rows after removing duplicates:")
print(len(df))
```

```python

import pandas as pd

df = pd.read_csv("sales.csv")

df["Revenue"] = df["Quantity"] * df["Price"]

print("Sales Data with Revenue:")
print(df)

print("\nTotal Revenue by Category:")
category_revenue = df.groupby("Category")["Revenue"].sum()
print(category_revenue)

print("\nTotal Revenue:")
print(df["Revenue"].sum())
```

```python

import pandas as pd

df = pd.read_csv("sales.csv")

df["Revenue"] = df["Quantity"] * df["Price"]

print("Sales sorted by Revenue (highest first):")
sorted_df = df.sort_values(by="Revenue", ascending=False)
print(sorted_df)

print("\nCorrelation Matrix:")
print(df[["Quantity", "Price", "Revenue"]].corr())
```
